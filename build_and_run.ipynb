```python
import os
from collections import defaultdict
from pathlib import Path
import pydicom

FIELDS = [
    "Modality",
    "SeriesDescription",
    "SeriesInstanceUID",
    "StudyInstanceUID",
    "ImageType",
    "ProtocolName",
    "SequenceName",
    "SequenceVariant",
    "ScanningSequence",
    "BodyPartExamined",
    "Manufacturer",
    "ManufacturerModelName",
    "MagneticFieldStrength",
]

def read_header(path: Path):
    try:
        ds = pydicom.dcmread(
            str(path),
            stop_before_pixels=True,
            force=True,
            specific_tags=FIELDS,
        )
        return ds
    except Exception:
        return None

def normalize_value(val):
    if val is None:
        return None
    if isinstance(val, (list, tuple)):
        return [str(v) for v in val]
    return str(val)

def scan_dir(root: Path):
    series_map = defaultdict(list)
    total_files = 0
    dicom_files = 0

    for dirpath, _, filenames in os.walk(root):
        for fname in filenames:
            total_files += 1
            if fname.endswith(":Zone.Identifier"):
                continue
            path = Path(dirpath) / fname
            ds = read_header(path)
            if ds is None:
                continue
            dicom_files += 1
            sid = getattr(ds, "SeriesInstanceUID", "UNKNOWN_SERIES")
            series_map[sid].append(ds)

    return total_files, dicom_files, series_map

def summarize_series(series_map):
    summaries = []
    for sid, dsets in series_map.items():
        ds = dsets[0]
        summary = {f: normalize_value(getattr(ds, f, None)) for f in FIELDS}
        summary["NumInstances"] = len(dsets)
        summaries.append(summary)
    summaries.sort(key=lambda x: (x.get("SeriesDescription") or "", -x["NumInstances"]))
    return summaries

def print_summary(title, total_files, dicom_files, summaries):
    print(f"\n=== {title} ===")
    print(f"Files scanned: {total_files}, DICOM files read: {dicom_files}")
    if not summaries:
        print("No DICOM series found.")
        return
    for s in summaries:
        print("-" * 80)
        print(f"SeriesInstanceUID: {s.get('SeriesInstanceUID')}")
        print(f"SeriesDescription: {s.get('SeriesDescription')}")
        print(f"ImageType: {s.get('ImageType')}")
        print(f"Modality: {s.get('Modality')}")
        print(f"ProtocolName: {s.get('ProtocolName')}")
        print(f"SequenceName: {s.get('SequenceName')}")
        print(f"SequenceVariant: {s.get('SequenceVariant')}")
        print(f"ScanningSequence: {s.get('ScanningSequence')}")
        print(f"BodyPartExamined: {s.get('BodyPartExamined')}")
        print(f"Manufacturer: {s.get('Manufacturer')}")
        print(f"Model: {s.get('ManufacturerModelName')}")
        print(f"B0 (T): {s.get('MagneticFieldStrength')}")
        print(f"StudyInstanceUID: {s.get('StudyInstanceUID')}")
        print(f"NumInstances: {s.get('NumInstances')}")

# Set directories to scan
paths = [
    "/home/anson/work/research-contributions/prostate-mri-lesion-seg/test-data/ProstateX-0000",
    # "/home/anson/work/research-contributions/prostate-mri-lesion-seg/test-data/sgh_data_patient_2",
]

for p in paths:
    root = Path(p).expanduser().resolve()
    total_files, dicom_files, series_map = scan_dir(root)
    summaries = summarize_series(series_map)
    print_summary(str(root), total_files, dicom_files, summaries)

print("\nIf you see 'No DICOM series found', run this on the original DICOM inputs under test-data.")
```

```python
"""
DICOM Metadata Compatibility Diagnostic
========================================
Checks all test-data cases for:
  1. (0020,0010) StudyID — required by rt_utils for RTSTRUCT generation
  2. Related de-identification tags for Dr. Anh / Jun Yan investigation
  3. Per-series slice counts to flag series too thin for volume reconstruction

Findings so far:
  - ProstateX:  StudyID tag IS present (empty string '') → rt_utils works
  - SGH data:   StudyID tag is COMPLETELY MISSING       → rt_utils crashes
  - Dr. Anh says (0020,0010) StudyID should be a hash of AccessionNumber
    if de-id went correctly. Its absence suggests the de-id tool stripped
    it instead of hashing it.
"""

import os
import pydicom
from collections import defaultdict
from pathlib import Path

TEST_ROOT = Path("test-data")

TAGS_TO_CHECK = [
    ("StudyID",                0x0020, 0x0010, "REQUIRED by rt_utils for RTSTRUCT"),
    ("AccessionNumber",        0x0008, 0x0050, "Should hash to StudyID per de-id"),
    ("PatientID",              0x0010, 0x0020, "De-id: should be hashed"),
    ("PatientName",            0x0010, 0x0010, "De-id: should be hashed"),
    ("StudyInstanceUID",       0x0020, 0x000D, "Study-level UID"),
    ("SeriesInstanceUID",      0x0020, 0x000E, "Series-level UID"),
    ("ImageType",              0x0008, 0x0008, "Critical for series selection rules"),
    ("SeriesDescription",      0x0008, 0x103E, "Critical for series selection rules"),
    ("InstitutionName",        0x0008, 0x0080, "De-id: should be removed/hashed"),
    ("ReferringPhysicianName", 0x0008, 0x0090, "De-id: should be removed/hashed"),
]

SEPARATOR = "=" * 100

cases = sorted([d for d in os.listdir(TEST_ROOT) if (TEST_ROOT / d).is_dir()])

# ─── Part 1: StudyID presence summary ───────────────────────────────────────
print(SEPARATOR)
print("PART 1: StudyID (0020,0010) Presence — Summary Across All Cases")
print(SEPARATOR)
print(f"{'Case':<25} {'StudyID Present?':>17} {'StudyID Value':<25} {'AccessionNumber':<25}")
print("-" * 100)

case_sample_files = {}
for case in cases:
    case_path = TEST_ROOT / case
    sample_file = None
    for dirpath, _, filenames in os.walk(case_path):
        for fn in sorted(filenames):
            if fn.lower().endswith(".dcm"):
                sample_file = os.path.join(dirpath, fn)
                break
        if sample_file:
            break
    if not sample_file:
        continue

    ds = pydicom.dcmread(sample_file, stop_before_pixels=True, force=True)
    case_sample_files[case] = (sample_file, ds)

    tag_present = (0x0020, 0x0010) in ds
    study_id = getattr(ds, "StudyID", None)
    accession = getattr(ds, "AccessionNumber", None)

    status = "YES" if tag_present else "** MISSING **"
    val_display = repr(study_id) if study_id is not None else "N/A (tag absent)"
    print(f"{case:<25} {status:>17} {val_display:<25} {accession or 'N/A':<25}")

# ─── Part 2: Full tag audit per case ────────────────────────────────────────
print(f"\n{SEPARATOR}")
print("PART 2: Full DICOM Tag Audit (one sample file per case)")
print(SEPARATOR)

for case in cases:
    if case not in case_sample_files:
        continue
    sample_file, ds = case_sample_files[case]
    fn = os.path.basename(sample_file)
    print(f"\n--- {case} (sample: {fn}) ---")
    print(f"  {'Tag':12s} {'Keyword':<30s} {'Present':>8s}  {'Value':<50s}  Note")
    print(f"  {'-'*12} {'-'*30} {'-'*8}  {'-'*50}  {'-'*30}")
    for name, g, e, note in TAGS_TO_CHECK:
        present = (g, e) in ds
        val = getattr(ds, name, None)
        val_str = repr(val) if val is not None else "N/A"
        if len(val_str) > 48:
            val_str = val_str[:45] + "..."
        flag = "YES" if present else "MISSING"
        alert = " <<<< ISSUE" if (not present and "REQUIRED" in note) else ""
        print(f"  ({g:04X},{e:04X}) {name:<30s} {flag:>8s}  {val_str:<50s}  {note}{alert}")

# ─── Part 3: Per-series slice counts ────────────────────────────────────────
print(f"\n{SEPARATOR}")
print("PART 3: Per-Series Slice Counts (flags series with < 3 slices)")
print(SEPARATOR)

MIN_SLICES_FOR_AFFINE = 2
MIN_SLICES_FOR_MODEL = 5

for case in cases:
    case_path = TEST_ROOT / case
    series_slices = defaultdict(int)
    series_desc = {}
    series_image_type = {}

    for dirpath, _, filenames in os.walk(case_path):
        for fn in filenames:
            if not fn.lower().endswith(".dcm"):
                continue
            fp = os.path.join(dirpath, fn)
            try:
                ds = pydicom.dcmread(fp, stop_before_pixels=True, force=True)
                uid = str(getattr(ds, "SeriesInstanceUID", "unknown"))
                series_slices[uid] += 1
                series_desc[uid] = str(getattr(ds, "SeriesDescription", "N/A"))
                it = getattr(ds, "ImageType", None)
                series_image_type[uid] = str(it) if it else "None"
            except Exception:
                pass

    total = sum(series_slices.values())
    print(f"\n--- {case} --- ({total} DCMs, {len(series_slices)} series)")
    print(f"  {'Slices':>6s}  {'Flag':>12s}  {'SeriesDescription':<55s}  ImageType")
    print(f"  {'-'*6}  {'-'*12}  {'-'*55}  {'-'*50}")

    for uid, count in sorted(series_slices.items(), key=lambda x: -x[1]):
        desc = series_desc.get(uid, "?")
        itype = series_image_type.get(uid, "?")
        if len(itype) > 48:
            itype = itype[:45] + "..."
        if count < MIN_SLICES_FOR_AFFINE:
            flag = "** CRASH **"
        elif count < MIN_SLICES_FOR_MODEL:
            flag = "LOW"
        else:
            flag = "OK"
        print(f"  {count:>6d}  {flag:>12s}  {desc:<55s}  {itype}")

# ─── Part 4: Recommendation for Dr. Anh ─────────────────────────────────────
print(f"\n{SEPARATOR}")
print("PART 4: Findings & Recommendation for Dr. Anh / Jun Yan")
print(SEPARATOR)
print("""
FINDING: (0020,0010) StudyID tag is COMPLETELY ABSENT from all SGH DICOM files.
         ProstateX data has StudyID present (as empty string '').

         The de-identification process for SGH data appears to have STRIPPED
         the StudyID tag entirely instead of replacing it with a hash of the
         AccessionNumber.

IMPACT:  rt_utils (used for RTSTRUCT generation) calls pydicom's FileDataset
         which expects StudyID to exist. When missing, it raises:
           'FileDataset' object has no attribute 'StudyID'

AFFECTED PATIENTS:
  - sgh_data_patient_1  (AccessionNumber: 8f4a1eb6f46266320a70)
  - sgh_data_patient_2  (AccessionNumber: 823319ef014cb44958c1)
  - sgh_data_patient_3  (AccessionNumber: 68bf25e98b00289cdb00)
  - sgh_data_patient_4  (AccessionNumber: eed80f7e38dd92b9cb9d)

ACTION NEEDED:
  1. Share any SGH DICOM study with Jun Yan to confirm this is a systemic
     de-id issue (the AccessionNumber values above can identify the studies).
  2. Jun Yan should fix the de-id pipeline to HASH StudyID rather than strip it.
  3. As a workaround, our app can inject a synthetic StudyID (hash of
     AccessionNumber) at runtime before RTSTRUCT generation.
""")
```
